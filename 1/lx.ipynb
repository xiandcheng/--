```python
from pathlib import Path
import json
import csv
import pandas as pd
```

```python

files_path = Path("./book")
my_dict = {"title":[],"author":[],"year":[],"price":[],"tags":[]}
for file in files_path.rglob("*.json"):

    with open(file,'r',encoding='utf8') as f:
        file_data = json.load(f)
    for k in file_data.keys():
        if k =='tags':
            if len(file_data[k])==0:
                my_dict[k].append("无")
            else:
               my_dict[k].append(file_data[k][0]) 
        else:
            my_dict[k].append(file_data[k])

my_dict
```

```python
df =pd.DataFrame(my_dict)
df
```

```python
df.to_csv("./data.csv",index = False)
```
